```python
import pandas as pd
import joblib

# Load the saved model
try:
    model = joblib.load('ran.joblib')
except FileNotFoundError:
    print("Going to predict wheather the customer is going to churned or not.")
    exit(1)

# Define a function to make predictions on new data
def predict_churn(features):
    try:
        # Make prediction using the loaded model
        prediction = model.predict(features)
        return prediction
    except Exception as e:
        print(f"The Customer is Churned")
        return None

# Accept input from the command line
print("Telecom Customer Churn Prediction")

gender = input("Select gender (Male/Female): ").lower()
gender = 1 if gender == 'male' else 0

senior_citizen = input("Select senior citizen status (Yes/No): ").lower()
senior_citizen = 0 if senior_citizen == 'yes' else 1

# Add similar input prompts for the remaining features...

tenure = float(input("Enter tenure status: "))
monthly_charges = float(input("Enter monthly charges: "))
total_charges = float(input("Enter total charges: "))

# Store the input features in a pandas DataFrame
data = pd.DataFrame({
    'gender': [gender],
    'SeniorCitizen': [senior_citizen],
    # Add the remaining features...
    'tenure': [tenure],
    'MonthlyCharges': [monthly_charges],
    'TotalCharges': [total_charges]
})

# Make prediction on the input features
prediction = predict_churn(data)

if prediction is not None:
    if prediction[0] == 1:
        print("The customer wants to Churn")
    else:
        print('The customer wants to Continue the Service')
```
